```python
# ============================================
# California Coastal Conditions Pipeline
# Stage 1 of 4: Clean the CalCOFI flat files
# Author: Brittany Blessie
# Description: Loads the CalCOFI Bottle and Cast CSVs, joins them on Cst_Cnt, and
#              applies labeled transformation steps to produce a clean coastal dataset.
# ============================================
```

```python
import pandas as pd
```

```python
cast = pd.read_csv("194903-202105_Cast.csv", low_memory=False)
```

```python
cast.shape
cast.head()
```

```python
cast.columns.tolist()
```

## Step 1: Select Cast Columns
Keep only the join key, decimal coordinates, date, time, and station ID from the Cast file. Drops the other 55 columns that the project does not use.

```python
# Step #1: Select only the columns needed from the Cast file.
# Keeps the join key, decimal coordinates, date/time, and station ID; drops the other 55 columns.
cast_clean = cast[["Cst_Cnt", "Lat_Dec", "Lon_Dec", "Date", "Time", "Sta_ID"]]
cast_clean.shape
```

```python
bottle = pd.read_csv("194903-202105_Bottle.csv", encoding="latin1", low_memory=False)
```

```python
bottle.shape
bottle.columns.tolist()
```

## Step 2: Select Bottle Columns
Keep only the six needed columns from the Bottle file: the join key, the bottle ID, and the four ocean measurements (depth, temperature, salinity, oxygen). Drops the other 56 columns.

```python
# Step #2: Keep only the six needed columns from the Bottle file and drop the other 56.
# These are the join key, the bottle ID, and the four ocean measurements (depth, temperature, salinity, oxygen).
bottle_clean = bottle[["Cst_Cnt", "Btl_Cnt", "Depthm", "T_degC", "Salnty", "O2ml_L"]]
bottle_clean.shape
```

## Step 3: Join Bottle and Cast on Cst_Cnt
Merge the Bottle and Cast tables on Cst_Cnt so every measurement carries the coordinates from its cast. This is a many-to-one join, since each cast has many bottle samples.

```python
# Step #3: Join the Bottle and Cast tables on Cst_Cnt so every measurement carries its cast's coordinates.
# This is a many-to-one merge: many bottle rows attach to one cast row that shares the same Cst_Cnt.
merged = bottle_clean.merge(cast_clean, on="Cst_Cnt", how="left")
merged.shape
```

```python
merged.isnull().sum()
```

## Step 4: Drop Rows Missing Measurements
Remove any row missing depth, temperature, salinity, or oxygen, since the analysis needs all four together. This drops about 204,000 incomplete rows and leaves complete records.

```python
# Step #4: Drop rows missing any of the four ocean measurements.
# The analysis needs temperature, salinity, oxygen, and depth together, so only complete rows are kept.
merged_clean = merged.dropna(subset=["Depthm", "T_degC", "Salnty", "O2ml_L"])
merged_clean.shape
```

```python
merged_clean.describe()
```

## Step 5: Remove Impossible Oxygen Values
Drop rows where oxygen is below zero. Negative oxygen is physically impossible, so these are placeholder or bad values that the missing-value step did not catch.

```python
# Step #5: Remove physically impossible oxygen values.
# Oxygen cannot be negative, so rows with O2ml_L below 0 are placeholder/bad data and are dropped.
merged_clean = merged_clean[merged_clean["O2ml_L"] >= 0]
merged_clean.shape
```

## Step 6: Reset the Index
Renumber the rows from zero after the drops so the final dataset has a clean, continuous index.

```python
# Step #6: Reset the row index after dropping rows so the final dataset is numbered cleanly from 0.
merged_clean = merged_clean.reset_index(drop=True)
merged_clean.shape
```

```python
# Final cleaned and merged dataset (first 20 rows)
merged_clean.head(20)
```

## Step 7: Rename Columns to Readable Names
Rename the abbreviated CalCOFI field names to clear labels with units, so the final dataset is understandable without a codebook. Columns like T_degC become Temperature_C and O2ml_L becomes Oxygen_mL_L.

```python
# Step #7: Rename columns to clear, human-readable names.
# CalCOFI uses abbreviated field names, so this makes the final dataset understandable without a codebook.
merged_clean = merged_clean.rename(columns={
    "Cst_Cnt": "Cast_ID",
    "Btl_Cnt": "Bottle_ID",
    "Depthm": "Depth_m",
    "T_degC": "Temperature_C",
    "Salnty": "Salinity_PSU",
    "O2ml_L": "Oxygen_mL_L",
    "Lat_Dec": "Latitude",
    "Lon_Dec": "Longitude",
    "Sta_ID": "Station_ID"
})
merged_clean.head(20)
```

## Data Notes and Limitations

The transformations I made to the CalCOFI flat file were as follows: I selected a small set of columns from the Bottle and Cast files, joined the two on Cst_Cnt so every measurement carried its cast coordinates, dropped rows missing any of the four measurements, removed five rows with negative oxygen values that are physically impossible, and reset the index. On legal and regulatory guidelines, CalCOFI data is public research data collected by Scripps, NOAA, and California Fish and Wildlife, published for open scientific use, so there are no licensing restrictions and no personal information is involved. The main risk created by these transformations is that the cleaned file is now a subset of the full record, since dropping about 204,000 incomplete rows means anyone who treated this version as the complete CalCOFI archive could misjudge how much data exists or how the coast was sampled. As for assumptions, the biggest is that a row needed all four measurements to be useful, since the analysis compares them together. I also treated the negative oxygen readings as bad data rather than real values, because oxygen cannot be negative, and I assumed the coordinates recorded for each cast apply to all the bottles collected during that cast, which is how CalCOFI structures its sampling. On sourcing and credibility, I downloaded the data directly from the official CalCOFI website and checked each column meaning against their published field descriptions. The data was acquired ethically, since it is public data intended for research and I obtained it from the original source. To mitigate the risks I identified, I documented every transformation and the number of rows it affected inside the notebook, kept the original raw files so the complete data can always be recovered, and will describe the cleaned dataset as a subset rather than the full archive whenever I present it.
